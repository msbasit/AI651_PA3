# Intel AI Boost NPU deployment
Training uses PyTorch XPU autograd. OpenVINO is used for NPU inference, with explicit NPU selection and no AUTO/CPU fallback. Static batch one avoids the NPU compiler striding limitation encountered at batch 16. Learned adjacency is frozen only for export after training. Each saved model is evaluated on every test window.

```python
from pathlib import Path
import sys, subprocess, json
from IPython.display import display, Image
ROOT=Path.cwd()
if ROOT.name=='notebooks': ROOT=ROOT.parent
assert (ROOT/'src').is_dir()
def run(script,*args):
    command=[sys.executable,str(ROOT/'src'/script),*args]
    log=ROOT/'results'/(script.replace('.py','')+('_'+args[-1] if args else '')+'.log')
    with log.open('w',encoding='utf-8') as f:
        result=subprocess.run(command,cwd=ROOT,stdout=f,stderr=subprocess.STDOUT)
    print(log.read_text(encoding='utf-8')[-16000:])
    result.check_returncode()
def show(name):
    display(Image(filename=str(ROOT/'figures'/name)))
```

```python
run('npu.py')
print((ROOT/'results'/'npu_metrics.json').read_text())
```
